## 1: Topic

I chose **transfer learning** because it is the most natural next step after training a CNN from scratch for scene classification. The Milestone 4 model learned all visual features directly from the Intel scene dataset. Transfer learning asks whether we can start from a model that already learned useful visual features from a much larger dataset, then adapt it to our six scene classes.

## 2: What is Transfer Learning

Transfer learning is a deep learning technique where a model trained on one large source task is reused as the starting point for a new, related target task. In computer vision, this often means taking a convolutional network such as MobileNetV2, ResNet, or EfficientNet that was trained on ImageNet, removing its original classification head, and attaching a smaller task-specific classifier. The early and middle layers of these networks usually learn general visual patterns such as edges, textures, corners, shapes, and object parts. Because those patterns are useful across many image tasks, the model does not need to relearn them from scratch.

A common workflow has two stages. First, the pretrained base is frozen and only the new classifier head is trained. This quickly adapts the model to the new labels while preserving the general visual representation. Second, some upper layers of the base model may be unfrozen and fine-tuned with a low learning rate. Fine-tuning lets the model adjust higher-level features to the target dataset without destroying useful pretrained weights.

Transfer learning is widely used in image classification, medical imaging, defect detection, wildlife monitoring, satellite imagery, and mobile vision systems. It is especially useful when the target dataset is moderate or small, because it can improve accuracy, reduce training time, and lower compute costs compared with training a deep model from scratch.

## 3: Connection to Previous Work

In Milestone 4, I trained a compact CNN from scratch to classify natural scenes into buildings, forest, glacier, mountain, sea, and street. Transfer learning could improve this workflow by replacing the custom convolution blocks with a pretrained MobileNetV2 or EfficientNet feature extractor. The data pipeline, evaluation metrics, confusion matrix, and TFLite deployment workflow would stay mostly the same, but the model would start with stronger visual features. This could improve test accuracy, reduce the number of epochs needed, and make the model more practical for a capstone project. The trade-off is that the model may become larger or inherit biases from the source dataset, so I would compare accuracy, model size, latency, and error patterns against the current scratch CNN.

## 4: Reflection

A major consequence of transfer learning is **hidden dataset bias**. A pretrained vision model carries assumptions from its source data: what environments, lighting conditions, camera qualities, geographic regions, and object appearances were common during pretraining. If those assumptions do not match the deployment context, the model can appear accurate in testing while failing for underrepresented places or image conditions. For this scene classifier, a model pretrained on common internet imagery might recognize urban streets or forests well but perform worse on rural roads, informal settlements, unusual architecture, snow glare, monsoon lighting, or low-cost mobile camera images.

In a real deployment, I would address this by evaluating performance across meaningful slices, not just overall accuracy. For example, I would track per-class recall, inspect confusion-matrix failures, and test images from different regions, seasons, lighting conditions, and camera sources. I would also use explanation tools such as Grad-CAM to check whether the model focuses on meaningful scene regions rather than shortcuts like sky color, watermarks, or camera artifacts. If the model is used in a consequential setting, the deployment should include confidence thresholds, human review for uncertain cases, dataset documentation, and periodic re-evaluation after new data is collected.

## Capstone Implication

For the capstone, transfer learning suggests a stronger modeling baseline: start with a pretrained mobile-friendly model, fine-tune it on the project dataset, and compare it against the scratch CNN from Milestone 4. The comparison should include accuracy, class-level errors, training time, exported model size, and inference latency so the final choice is justified by both performance and deployment constraints.
